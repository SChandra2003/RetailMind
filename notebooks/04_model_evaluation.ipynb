```python
import pandas as pd
import numpy as np

import joblib

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)
```

```python
import pandas as pd
import joblib
from pathlib import Path

BASE_DIR = Path.cwd()

if not (BASE_DIR / "models").exists():
    BASE_DIR = BASE_DIR.parent
```

```python
MODEL_PATH = BASE_DIR / "models" / "churn_pipeline.joblib"

model = joblib.load(MODEL_PATH)
```

```python
TEST_DATA_PATH = BASE_DIR / "data" / "processed" / "test_data.csv"

test_data = pd.read_csv(TEST_DATA_PATH)

X_test = test_data.drop(columns=["Churn"])
y_test = test_data["Churn"]
```

```python
y_pred = model.predict(X_test)

y_prob = model.predict_proba(X_test)[:, 1]
```

```python
metrics = {
    "accuracy": accuracy_score(y_test, y_pred),
    "precision": precision_score(y_test, y_pred),
    "recall": recall_score(y_test, y_pred),
    "f1_score": f1_score(y_test, y_pred),
    "roc_auc": roc_auc_score(y_test, y_prob)
}

metrics
```

```python
for metric, value in metrics.items():
    print(f"{metric}: {value:.4f}")
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["No Churn", "Churn"]
    )
)
```

```python
import json
from pathlib import Path

BASE_DIR = Path.cwd()

if not (BASE_DIR / "reports").exists():
    BASE_DIR = BASE_DIR.parent

REPORT_PATH = BASE_DIR / "reports" / "baseline_metrics.json"

REPORT_PATH.parent.mkdir(parents=True, exist_ok=True)

with open(REPORT_PATH, "w") as f:
    json.dump(metrics, f, indent=4)

print("Evaluation metrics saved successfully!")
```
